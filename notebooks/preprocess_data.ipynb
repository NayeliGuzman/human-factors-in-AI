## Pre-process, create features, and annotate ChatGPT and Gemini paper review data
First we must label and annotate the data.
We add labels to show primary are of research and which model was used. We also flag the following features of the data:

-Figure on the first/second page [‘first_fig’]

-Number of figures, the sum of tables and figures [‘num_figures’]

-Number of references [‘num_references’]

-Number of supplementary material pages [‘num_supp_materials’]

-Research area (or keywords) [‘primary_area’] 

-The length of different sections 

-Use of mathematical equations/formulas as a proof  [‘uses_equation’] (1 - yes, 0 - no)

-Dataset sizes mentioned [‘dataset’]

```python
# Read the JSON data from the file and add titles
df_gpt = pd.read_json("../json_files/chatgpt_reviews.json")
df_gemini = pd.read_json("../json_files/gemini_reviews.json")
df_gpt['model'] = 'ChatGPT'
df_gemini['model'] = 'Gemini'
df_gpt = df_gpt[["model"] + list(df_gpt.columns[:-1])]
df_gemini = df_gemini[["model"] + list(df_gemini.columns[:-1])]
titles = np.array([   
'Learning to grok: Emergence of in-context learning and skill composition in modular arithmetic tasks',
'Trading Place for Space: Increasing Location Resolution Reduces Contextual Capacity in Hippocampal Codes',
'Visual Autoregressive Modeling: Scalable Image Generation via Next-Scale Prediction',
'Cracking the Code of Juxtaposition: Can AI Models Understand the Humorous Contradictions',
'Human Expertise in Algorithmic Prediction',   
'Learning diffusion at lightspeed',  
'Weisfeiler and Leman Go Loopy: A New Hierarchy for Graph Representational Learning',    
'Unlocking the Capabilities of Thought: A Reasoning Boundary Framework to Quantify and Optimize Chain-of-Thought',
'Reinforcement Learning Under Latent Dynamics: Toward Statistical and Algorithmic Modularity',
'Generalization Error Bounds for Two-stage Recommender Systems with Tree Structure',
'Aligner: Efficient Alignment by Learning to Correct',   
'Questioning the Survey Responses of Large Language Models', 
'Stochastic Taylor Derivative Estimator: Efficient amortization for arbitrary differential operators',
'Do Finetti: On Causal Effects for Exchangeable Data',
'LLM Evaluators Recognize and Favor Their Own Generations',
'Span-Based Optimal Sample Complexity for Weakly Communicating and General Average Reward MDPs',
'Learning Formal Mathematics From Intrinsic Motivation',
'Identification and Estimation of the Bi-Directional MR with Some Invalid Instruments',   
'Maximum Entropy Inverse Reinforcement Learning of Diffusion Models with Energy-Based Models',    
'Improving Environment Novelty Quantification for Effective Unsupervised Environment Design',   
'Enhancing Preference-based Linear Bandits via Human Response Time',
'Scale Equivariant Graph Metanetworks',
'CAT3D: Create Anything in 3D with Multi-View Diffusion Models',
'Stylus: Automatic Adapter Selection for Diffusion Models',    
'The Sample-Communication Complexity Trade-off in Federated Q-Learning'
])   
    

df_columns = pd.DataFrame(titles, columns=['titles'])

# Ensure the number of rows in df_columns matches df_gpt & df_gemini
df_columns = df_columns.iloc[:len(df_gpt)].reset_index(drop=True)

# Insert "titles" as the first column in df_gpt and df_gemini
df_gpt = pd.concat([df_columns, df_gpt], axis=1)
df_gemini = pd.concat([df_columns, df_gemini], axis=1)
```
